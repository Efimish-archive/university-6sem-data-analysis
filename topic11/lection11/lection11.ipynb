```python
# Подготовка зависимостей
import numpy as np
import scipy.stats as stats
```

### 1. t-статистика и t-распределение: понятие и назначение.

**t-статистика** — это статистика, которая используется вместо z-оценки для проверки гипотез о среднем для генеральной совокупности, когда неизвестно стандартное отклонение для генеральной совокупности.

Значение t-статистики вычисляется практически так же, как и для z-оценки: разница заключается лишь в том, что в знаменателе ставится оценка для стандартной ошибки:

$$t = \frac{M - \mu}{s_m}$$

Для вычисления оценки стандартной ошибки необходимо извлечь корень из частного дисперсии для выборки и объема выборки:

$$s_m = \sqrt{\frac{s^2}{n}}$$

Основное отличие в том, что в z-оценке используется стандартное отклонение для генеральной совокупности, а в t-статистике — для выборки.

**t-распределение** — это набор всех значений t-статистик, вычисленных для всех возможных случайных выборок определенного размера. t-распределение стремится к нормальному распределению с увеличением выборки.

Количество степеней свободы описывает вариативность выборки и равняется объему выборки, из которого вычли единицу. Чем больше степень свободы — тем ближе друг к другу значения стандартного отклонения для генеральной совокупности и выборки и, следовательно, тем ближе значения z-оценки и t-статистики.

### 2. Одновыборочный t-тест: применение и алгоритм проверки гипотез.

Одновыборочный t-тест применяется, когда мы хотим проверить, что среднее значение какого-то признака для генеральной совокупности равно определённому числу, имея только выборку.

**Алгоритм проверки гипотезы о равенстве среднего числу:**

1. Находим наблюдаемое значение t-статистики по формуле:

$$t_{\text{наблюдаемое}} = \frac{M - \mu}{s_m}$$

2. По уровню значимости и количеству степеней свободы определяем критические значения t-статистики по таблице.

3. Если наблюдаемое значение не попадает в критическую область — не отвергаем нулевую гипотезу. Если наблюдаемое значение попадает в критическую область — отвергаем нулевую гипотезу в пользу альтернативной.

### 3. Реализация одновыборочного t-теста в Python.

Провести одновыборочный t-тест в Python можно с помощью функции `ttest_1samp()` из библиотеки `scipy.stats`.

Например, попытаемся понять на основе выборки, правда ли, что средний рост всех мужчин в России составляет 182 см.

**Нулевая гипотеза H₀:** рост действительно составляет 182 см.

**Альтернативная гипотеза H₁:** рост не равен 182 см.

```python
np.random.seed(1)
height = list(np.round(np.random.normal(180, 10, 1000)))

t_statistic, p_value = stats.ttest_1samp(a=height, popmean=182)
print(t_statistic, p_value)

if p_value < 0.05:  # уровень значимости 0,05 или 5%
    print("Есть основания отвергнуть нулевую гипотезу")
else:
    print("Нет оснований отвергнуть нулевую гипотезу")
```

Найти критическое значение t-статистики в Python можно с помощью функции `scipy.stats.t.ppf(q, df)`.

Существует три различных типа проверки гипотез:
- двусторонний тест: альтернативная гипотеза содержит знак «≠»;
- левосторонний тест: альтернативная гипотеза содержит знак «<»;
- правосторонний тест: альтернативная гипотеза содержит знак «>».

Рассчитаем критическое значение t-статистики для двустороннего теста с уровнем значимости α=0,05 и числом степеней свободы 999 (в выборке 1000 объектов):

```python
# Критическое значение для двустороннего теста
t_critical = stats.t.ppf(q=1-.05/2, df=999)
print(f"Критическое значение t: {t_critical}")
print(f"Наблюдаемое значение t-статистики: {t_statistic:.4f}")
print(f"Наблюдаемое значение {t_statistic:.2f} < -{t_critical:.2f}, поэтому нулевая гипотеза отвергается.")
```

### 4. Решите 2 способами (проведя свои расчёты в Python и с помощью библиотеки scipy.stats) задачу: даны баллы за экзамен по английскому языку, проверьте гипотезу о равенстве среднего балла всех учеников значению 6,5. Сравните результаты решений.
   Баллы: 8, 6, 4, 6, 8, 9, 7, 8.

**Нулевая гипотеза H₀:** µ = 6,5

**Альтернативная гипотеза H₁:** µ ≠ 6,5

```python
# Способ 1: ручные расчёты
scores = np.array([8, 6, 4, 6, 8, 9, 7, 8])
mu_0 = 6.5

n = len(scores)
M = np.mean(scores)
s = scores.std(ddof=1)          # стандартное отклонение выборки
sm = s / np.sqrt(n)             # оценка стандартной ошибки

t_obs = (M - mu_0) / sm
df = n - 1

# p-value для двустороннего теста
p_value_manual = 2 * (1 - stats.t.cdf(abs(t_obs), df))

print("=== Способ 1: ручные расчёты ===")
print(f"Выборочное среднее M = {M}")
print(f"Стандартное отклонение s = {round(s, 4)}")
print(f"Стандартная ошибка sm = {round(sm, 4)}")
print(f"Наблюдаемая t-статистика = {round(t_obs, 4)}")
print(f"Степени свободы df = {df}")
print(f"p-value = {round(p_value_manual, 4)}")

if p_value_manual < 0.05:
    print("Есть основания отвергнуть нулевую гипотезу")
else:
    print("Нет оснований отвергнуть нулевую гипотезу")
```

```python
# Способ 2: scipy.stats
t_stat_lib, p_value_lib = stats.ttest_1samp(a=scores, popmean=mu_0)

print("=== Способ 2: scipy.stats ===")
print(f"Наблюдаемая t-статистика = {round(t_stat_lib, 4)}")
print(f"p-value = {round(p_value_lib, 4)}")

if p_value_lib < 0.05:
    print("Есть основания отвергнуть нулевую гипотезу")
else:
    print("Нет оснований отвергнуть нулевую гипотезу")

print("\n=== Сравнение результатов ===")
print(f"t-статистика (ручной расчёт): {round(t_obs, 4)}, t-статистика (scipy): {round(t_stat_lib, 4)}")
print(f"p-value (ручной расчёт): {round(p_value_manual, 4)}, p-value (scipy): {round(p_value_lib, 4)}")
print("Результаты совпадают. Нулевая гипотеза не отвергается: средний балл приблизительно равен 6,5.")
```

### 5. Парный t-тест: применение и алгоритм проверки гипотез.

Парный t-тест применяется для проверки гипотезы о равенстве средних для парных (или связных) выборок. Такие выборки особенны тем, что каждому объекту из первой однозначно соответствует объект из другой. Например, это могут быть пары близнецов, а также такие выборки часто получаются при повторных измерениях.

Нулевая гипотеза заключается в нулевой разнице, а альтернативная — в том, что разница не равна нулю:

H₀: MD = 0  
H₁: MD ≠ 0

**Алгоритм проверки гипотезы о равенстве средних в случае связных выборок:**

1. Находим разницы D между вторым значением и первым для каждого объекта.

2. Находим наблюдаемое значение t-статистики для полученных разниц по формуле:

$$t_{\text{наблюдаемое}} = \frac{M_D - \mu_D}{s_{M_D}}$$

3. По уровню значимости и количеству степеней свободы определяем критические значения t-статистики по таблице.

4. Если наблюдаемое значение не попадает в критическую область — не отвергаем нулевую гипотезу. Если наблюдаемое значение попадает в критическую область — отвергаем нулевую гипотезу в пользу альтернативной.

### 6. Реализация парного t-теста в Python

Провести парный t-тест в Python можно с помощью функции `ttest_rel()` из библиотеки `scipy.stats`.

Например, проверим, значительно ли влияет прохождение подготовительного курса на успеваемость студента на экзамене. У нас есть 15 студентов, которые проходят предварительный тест. Затем каждый из студентов участвует в подготовительном курсе в течение двух недель. Затем учащиеся пересдают тест аналогичной сложности.

**H₀:** Средние баллы до и после теста равны.  
**H₁:** Средние баллы до и после теста не равны.

```python
pre = [88, 82, 84, 93, 75, 78, 84, 87, 95, 91, 83, 89, 77, 68, 91]  # результаты до прохождения курса
post = [91, 84, 88, 90, 79, 80, 88, 90, 90, 96, 88, 89, 81, 74, 92]  # результаты после

t_stat, p_value = stats.ttest_rel(pre, post)
print(f"t-статистика: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("Отвергаем нулевую гипотезу: средние баллы до и после теста различаются.")
else:
    print("Нет оснований отвергнуть нулевую гипотезу: средние баллы до и после теста равны.")
```

### 7. Решите 2 способами (проведя свои расчёты в Python и с помощью библиотеки scipy.stats) задачу: даны результаты экзаменов студентов по английскому языку до прохождения спец. курсов и после. Проверьте гипотезу: действительно ли курсы помогли улучшить знания. Сравните результаты решений.
   Баллы до курсов: 92, 98, 64, 70, 72, 81, 89, 85.
   Баллы после курсов: 96, 100, 70, 73, 79, 90, 90, 93.

**Нулевая гипотеза H₀:** MD = 0 (курсы не повлияли на знания)

**Альтернативная гипотеза H₁:** MD ≠ 0 (курсы повлияли на знания)

```python
# Способ 1: ручные расчёты
before = np.array([92, 98, 64, 70, 72, 81, 89, 85])
after  = np.array([96, 100, 70, 73, 79, 90, 90, 93])

D = after - before          # разницы: после минус до
MD = np.mean(D)             # среднее разниц
mu_D = 0                    # нулевая гипотеза: средняя разница = 0

n = len(D)
s_D = D.std(ddof=1)
sm_D = s_D / np.sqrt(n)

t_obs = (MD - mu_D) / sm_D
df = n - 1
p_value_manual = 2 * (1 - stats.t.cdf(abs(t_obs), df))

print("=== Способ 1: ручные расчёты ===")
print(f"Разницы D: {D}")
print(f"Среднее разниц MD = {MD}")
print(f"Стандартное отклонение разниц s_D = {round(s_D, 4)}")
print(f"Стандартная ошибка sm_D = {round(sm_D, 4)}")
print(f"Наблюдаемая t-статистика = {round(t_obs, 4)}")
print(f"Степени свободы df = {df}")
print(f"p-value = {round(p_value_manual, 4)}")

if p_value_manual < 0.05:
    print("Есть основания отвергнуть нулевую гипотезу: курсы помогли улучшить знания.")
else:
    print("Нет оснований отвергнуть нулевую гипотезу.")
```

```python
# Способ 2: scipy.stats
t_stat_lib, p_value_lib = stats.ttest_rel(before, after)

print("=== Способ 2: scipy.stats ===")
print(f"Наблюдаемая t-статистика = {round(t_stat_lib, 4)}")
print(f"p-value = {round(p_value_lib, 4)}")

if p_value_lib < 0.05:
    print("Есть основания отвергнуть нулевую гипотезу: курсы помогли улучшить знания.")
else:
    print("Нет оснований отвергнуть нулевую гипотезу.")

print("\n=== Сравнение результатов ===")
print(f"t-статистика (ручной расчёт): {round(t_obs, 4)}, t-статистика (scipy): {round(t_stat_lib, 4)}")
print(f"p-value (ручной расчёт): {round(p_value_manual, 4)}, p-value (scipy): {round(p_value_lib, 4)}")
print("Результаты совпадают. Нулевая гипотеза отвергается: курсы действительно помогли улучшить знания.")
```

### 8. Двухвыборочный t-тест: применение и алгоритм проверки гипотез.

Двухвыборочный t-тест используется для проверки гипотез для независимых выборок — когда сравниваются значения для двух выборок, которые никак не связаны друг с другом.

Нулевая гипотеза заключается в том, что средние для двух выборок равны:  
H₀: µ₁ = µ₂

Альтернативная — что средние не равны:  
H₁: µ₁ ≠ µ₂

**Алгоритм проверки гипотезы:**

1. Находим средние для обеих выборок.

2. Находим объединённое стандартное отклонение:

$$s_{M_1 - M_2} = \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}$$

3. Вычисляем значение наблюдаемой t-статистики:

$$t = \frac{(M_1 - M_2) - (\mu_1 - \mu_2)}{s_{M_1 - M_2}}$$

4. По уровню значимости и количеству степеней свободы определяем критические значения t-статистики по таблице.

5. Если наблюдаемое значение не попадает в критическую область — не отвергаем нулевую гипотезу. Если наблюдаемое значение попадает в критическую область — отвергаем нулевую гипотезу в пользу альтернативной.

Если дисперсии отличаются, используется поправка Уэлча–Саттертуэйта для вычисления количества степеней свободы:

$$df = \frac{\left(\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}\right)^2}{\frac{1}{n_1-1}\left(\frac{s_1^2}{n_1}\right)^2 + \frac{1}{n_2-1}\left(\frac{s_2^2}{n_2}\right)^2}$$

### 9. Реализация двухвыборочного t-теста в Python.

Провести двухвыборочный t-тест в Python можно с помощью функции `ttest_ind(a, b, equal_var=True)` из библиотеки `scipy.stats`.

Например, исследователи хотят знать, имеют ли два разных вида растений одинаковую среднюю высоту. Чтобы проверить это, они собирают простую случайную выборку из 20 растений каждого вида.

**H₀:** µ₁ = µ₂ (средние значения высоты двух видов растений равны).

**H₁:** µ₁ ≠ µ₂ (средние значения высоты двух видов растений не равны).

```python
group1 = np.array([14, 15, 15, 16, 13, 8, 14, 17, 16, 14, 19, 20, 21, 15, 15, 16, 16, 13, 14, 12])
group2 = np.array([15, 17, 14, 17, 14, 8, 12, 19, 19, 14, 17, 22, 24, 16, 13, 16, 13, 18, 15, 13])

# Проверка соотношения дисперсий
var1 = np.var(group1, ddof=1)
var2 = np.var(group2, ddof=1)
print(f"Дисперсии: {round(var1, 4)}, {round(var2, 4)}")
ratio = (np.var(group2, ddof=1) / np.var(group1, ddof=1)).round(2)
print(f"Отношение большей дисперсии к меньшей: {ratio}")
# Отношение < 4, значит можно предположить равенство дисперсий

t_stat, p_value = stats.ttest_ind(a=group1, b=group2, equal_var=True)
print(f"\nt-статистика: {round(t_stat, 4)}")
print(f"p-value: {round(p_value, 5)}")

if p_value < 0.05:
    print("Отвергаем нулевую гипотезу: средняя высота растений различается.")
else:
    print("Нет оснований отвергнуть нулевую гипотезу: средняя высота растений одинакова.")
```

### 10. Решите 2 способами (проведя свои расчёты в Python и с помощью библиотеки scipy.stats) задачу: даны данные для 2 школ английского языка. Проверьте гипотезу: студенты в среднем сдают экзамен одинаково. Сравните результаты решений.

Баллы для 1 школы: 5, 2, 7, 7, 6, 5, 4, 4.\
Баллы для 2 школы: 7, 9, 6, 9, 6, 8, 4, 7.

**Нулевая гипотеза H₀:** µ₁ = µ₂ (студенты в среднем сдают экзамен одинаково)

**Альтернативная гипотеза H₁:** µ₁ ≠ µ₂ (студенты сдают экзамен по-разному)

```python
# Способ 1: ручные расчёты
school_1 = np.array([5, 2, 7, 7, 6, 5, 4, 4])
school_2 = np.array([7, 9, 6, 9, 6, 8, 4, 7])

M1 = np.mean(school_1)
M2 = np.mean(school_2)
n1, n2 = len(school_1), len(school_2)

s1_sq = school_1.var(ddof=1)   # дисперсия 1-й выборки
s2_sq = school_2.var(ddof=1)   # дисперсия 2-й выборки

# Объединённое стандартное отклонение
sm = np.sqrt(s1_sq / n1 + s2_sq / n2)

# Наблюдаемая t-статистика (µ₁ - µ₂ = 0 по нулевой гипотезе)
t_obs = ((M1 - M2) - 0) / sm
df = n1 + n2 - 2

p_value_manual = 2 * (1 - stats.t.cdf(abs(t_obs), df))

print("=== Способ 1: ручные расчёты ===")
print(f"M1 = {M1}, M2 = {M2}")
print(f"Дисперсии: s1² = {round(s1_sq, 4)}, s2² = {round(s2_sq, 4)}")
print(f"Объединённое стандартное отклонение sm = {round(sm, 4)}")
print(f"Наблюдаемая t-статистика = {round(t_obs, 4)}")
print(f"Степени свободы df = {df}")
print(f"p-value = {round(p_value_manual, 4)}")

if p_value_manual < 0.05:
    print("Есть основания отвергнуть нулевую гипотезу: студенты сдают экзамен по-разному.")
else:
    print("Нет оснований отвергнуть нулевую гипотезу.")
```

```python
# Способ 2: scipy.stats
t_stat_lib, p_value_lib = stats.ttest_ind(school_1, school_2, equal_var=True)

print("=== Способ 2: scipy.stats ===")
print(f"Наблюдаемая t-статистика = {round(t_stat_lib, 4)}")
print(f"p-value = {round(p_value_lib, 4)}")

if p_value_lib < 0.05:
    print("Есть основания отвергнуть нулевую гипотезу: студенты сдают экзамен по-разному.")
else:
    print("Нет оснований отвергнуть нулевую гипотезу.")

print("\n=== Сравнение результатов ===")
print(f"t-статистика (ручной расчёт): {round(t_obs, 4)}, t-статистика (scipy): {round(t_stat_lib, 4)}")
print(f"p-value (ручной расчёт): {round(p_value_manual, 4)}, p-value (scipy): {round(p_value_lib, 4)}")
print("Результаты совпадают. Нулевая гипотеза отвергается: студенты двух школ сдают экзамен по-разному.")
```
